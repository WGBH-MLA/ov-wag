```python
from aapb_collections.models import AAPBCollection
from aapb_exhibits.models import AAPBExhibit
from ov_collections.models import OpenVaultCollection
from exhibits.models import OpenVaultExhibit
```

```python
# from wagtail.models import Page
aapb = Page.objects.get(title='AAPB')

for page in aapb.get_children():
    if isinstance(page.specific, AAPBCollection):
        page.aapbcollection.special_collections = page.slug
        page.aapbcollection.specific.save_revision().publish()
    elif isinstance(page.specific, AAPBExhibit):
        page.aapbexhibit.special_collections = page.slug
        page.aapbexhibit.specific.save_revision().publish()
```

```python
ov = Page.objects.get(title='OV')

for page in ov.get_children():
    if isinstance(page.specific, OpenVaultCollection):
        page.openvaultcollection.special_collections = page.slug
        page.openvaultcollection.specific.save_revision().publish()
    elif isinstance(page.specific, OpenVaultExhibit):
        page.openvaultexhibit.special_collections = page.slug
        page.openvaultexhibit.specific.save_revision().publish()
```
