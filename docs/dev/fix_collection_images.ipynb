# Fix AAPBCollections `cover_image`

This is a quick data fix for migrating all collection `hero_images` to `cover_images`

This shouldn't have to be run again, but is provided in case it is useful!

```python
from aapb_collections import AAPBCollection

for collection in AAPBCollection.objects.all():
    if not collection.hero_image and collection.cover_image:
        continue
    collection.cover_image = collection.hero_image
    collection.hero_image = None
    collection.save_revision().publish()
```
